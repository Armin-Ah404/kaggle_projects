        Provides a general overview of the dataset and summarizes the key information needed for the initial analysis.

<p style="

">
Together, these functions provide a quick assessment of the dataset's
<strong>structure, completeness, and overall quality</strong> before proceeding with the EDA.
</p>

```python
def data_overview(df):

    variables = []
    count = []
    missing_value = []
    dtype = []
    percentage_missing = []
    unique_value = []
    maximum = []
    minimum = []
    average = []
    median= []

    for features in df.columns :

        variables.append(features)
        count.append(len(df[features]))
        missing_value.append(df[features].isna().sum())
        dtype.append(df[features].dtype)
        percentage_missing.append(df[features].isna().sum() / len(df) * 100)
        unique_value.append(df[features].nunique())
        maximum.append(df[features].max())
        minimum.append(df[features].min())
        average.append(df[features].mean())
        median.append(df[features].median())


    data_frame = pd.DataFrame({
        'Features' : variables,
        'Dtype' : dtype,
        'Total amount' : count,
        'Unique count' : unique_value,
        'Missing value' : missing_value,
        'Percentage of missing' : percentage_missing,
        'Minimum' : minimum,
        'Average' : average,
        'Median' : median,
        'Maximum' : maximum,

    })

    print(f"Total number of rows in data: {df.shape[0]}")
    print(f"Total number of columns in data: {df.shape[1]}")
    print(f'Shape: {df.shape}')
    
    return data_frame.style.background_gradient(cmap = "YlOrRd", axis=None)
```

📊 Checking for Outliers Using the IQR Method
</h2>

<p style="
    margin-bottom: 0;
    color: #fee2e2;
">
In this section, the <strong>Interquartile Range (IQR)</strong> method is used
to identify potential outliers in the numerical features of the heart disease dataset.
This helps us assess the distribution of the data and detect observations that
fall unusually far from the central range.
</p>

</div>

```python
def IQR(df):

    outlier = pd.DataFrame(index=df.index)
    
    for col in df.columns:
        
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outlier[col] = ((df[col] < lower) | (df[col] > upper))

    return outlier
```

```python
fig, ax = plt.subplots(1, 3,figsize=(20, 8), dpi=400)

outlier = IQR(df[numerical])

outlier_columns = [col for col in outlier.columns[:-1]]
disease = []
non_disease = []

values = np.array(disease) + np.array(non_disease)
labels = outlier_columns

for col in outlier_columns:
    disease.append(((outlier[col]) & (df['target']==1)).sum())
    non_disease.append(((outlier[col]) & (df['target']==0)).sum())

ax[0].bar(
    outlier_columns, 
    disease,
    color='darkorange',
    label='Disease(1)'
)
ax[0].bar(
    outlier_columns, 
    non_disease,
    color='royalblue',
    label='No disease(0)',
    bottom=disease
)


for container in ax[0].containers:
    ax[0].bar_label(container, padding=1)

ax[0].legend()

sns.boxplot(
    df[numerical[:-1]],
    ax=ax[1]
)

sns.violinplot(
    df[numerical[:-1]], 
    ax=ax[2]
)

plt.suptitle('Checking for outliers', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```
