📊 Dataset Structure
</h3>

<p style="font-size:13px; margin:7px 0;">
The function also reports the overall <b>number of rows</b>,
<b>number of columns</b>, and the complete <b>dataset shape</b>,
providing a quick summary of its overall structure.
</p>

<p style="font-size:13px; margin:7px 0;">
The calculated information is returned as a separate
<b>DataFrame</b>, making it easier to inspect data types,

in a single structured table.
</p>

<div style="
    background:#211919;
    border:1px solid #ef4444;
    border-radius:9px;
    padding:13px 15px;
    margin-top:16px;
">

<p style="
    font-size:13px;
    margin:0;
">
🧠 <b>Why is this useful?</b><br>
The <b>data_overview()</b> function serves as an initial
<b>data profiling step</b>. It helps identify the structure,
quality, and statistical characteristics of the dataset before
moving on to <b>data cleaning, preprocessing, visualization,
and further analysis</b>.
</p>

</div>

</div>

```python
def data_overview(df):

    variables = []
    count = []
    missing_value = []
    dtype = []
    percentage_missing = []
    unique_value = []
    maximum = []
    minimum = []
    average = []
    median= []

    for features in df.columns :

        variables.append(features)
        count.append(len(df[features]))
        missing_value.append(df[features].isna().sum())
        dtype.append(df[features].dtype)
        percentage_missing.append(df[features].isna().sum() / len(df) * 100)
        unique_value.append(df[features].nunique())
        maximum.append(df[features].max())
        minimum.append(df[features].min())
        average.append(df[features].mean())
        median.append(df[features].median())


    data_frame = pd.DataFrame({
        'Features' : variables,
        'Dtype' : dtype,
        'Total amount' : count,
        'Unique count' : unique_value,
        'Missing value' : missing_value,
        'Percentage of missing' : percentage_missing,
        'Minimum' : minimum,
        'Average' : average,
        'Median' : median,
        'Maximum' : maximum,

    })

    print(f"Total number of rows in data: {df.shape[0]}")
    print(f"Total number of numerical columns in data: {df.shape[1]}")
    print(f'Shape: {df.shape}')
    
    return data_frame
```

```python
def outlier_analysis(data, features, multiplier=1.5):

    results = []

    for feature in features:

        series = data[feature]
        
        Q1 = series.quantile(0.25)
        Q3 = series.quantile(0.75)
        
        iqr = Q3 - Q1

        lower_bound = Q1 - multiplier * iqr
        upper_bound = Q3 + multiplier * iqr


        mask = (series < lower_bound) | (series > upper_bound)

        outlier_count = mask.sum()
        total_count = len(series)

        results.append({
            'Feature': feature,
            'Q1': Q1,
            'Q3': Q3,
            'IQR': iqr,
            'Lower Bound': lower_bound,
            'Upper Bound': upper_bound,
            'Outlier Count': outlier_count,
            'Outlier %': (outlier_count / total_count) * 100
        })

    outlier = pd.DataFrame(results).sort_values('Outlier %', ascending=False).reset_index(drop=True)
    return (outlier)
```

```python
numeric = [f for f in numerical_features if f != 'Year']
outlier_features = outlier_analysis(
    df,
    numeric
)
```

🔄 Cross-Validation and Fold-Level Error Analysis
</h2>

<p style="font-size:14px; margin:7px 0;">
In this section, a reusable <b>5-Fold Cross-Validation function</b>
is created to evaluate the performance and stability of the
<b>Linear Regression</b> model across different subsets of the
training data.
</p>

<p style="font-size:14px; margin:7px 0;">
The function divides the data into <b>five folds</b>. In each
iteration, four folds are used for training and the remaining
fold is used for validation. This process is repeated until every
sample has been used for validation once.
</p>

<h3 style="
    color:#f87171;
    margin:19px 0 9px 0;
    font-size:17px;
">
⚙️ Cross-Validation Process
</h3>

<p style="font-size:14px; margin:7px 0 8px 0;">
For each fold, the function performs the following steps:
</p>

<ol style="
    font-size:13px;
    margin:7px 0 13px 18px;
    padding-left:12px;
">

<li style="margin:6px 0;">
The training and validation indexes are generated using
<b><code>KFold</code></b>.
</li>

<li style="margin:6px 0;">
The corresponding training and validation samples are separated
using their indexes.
</li>

<li style="margin:6px 0;">
A new <b><code>StandardScaler</code></b> is created for the
current fold.
</li>

<li style="margin:6px 0;">
The scaler is fitted only on the training portion and is then
used to transform the validation data.
</li>

<li style="margin:6px 0;">
The <b>Linear Regression</b> model is trained using the scaled
training data.
</li>

<li style="margin:6px 0;">
Predictions are generated for the validation samples.
</li>

<li style="margin:6px 0;">
The <b>R² score</b> is calculated and stored for the current fold.
</li>

<li style="margin:6px 0;">
The original indexes of the validation samples are recovered
to connect the predictions back to the original dataset.
</li>

<li style="margin:6px 0;">
The actual values, predicted values, and absolute prediction
errors are added to a temporary DataFrame.
</li>

<li style="margin:6px 0;">
The <b>10 samples with the largest absolute errors</b> are
identified and displayed for further investigation.
</li>

</ol>

<h3 style="
    color:#f87171;
    margin:19px 0 9px 0;
    font-size:17px;
">
🛡️ Preventing Data Leakage
</h3>

<p style="font-size:14px; margin:7px 0;">
To prevent <b>data leakage</b>, the <code>StandardScaler</code>
is fitted separately inside every fold. This means that the
scaling parameters are calculated only from the training portion
of that fold and are never learned from the validation data.
</p>

<div style="
    background:#211919;
    border:1px solid #ef4444;
    border-radius:9px;
    padding:13px 15px;
    margin:14px 0 18px 0;
">

<p style="font-size:13px; margin:0;">
⚠️ <b>Important:</b> The validation fold is used only to evaluate
the model. It does not influence the scaling parameters or the
model training process.
</p>

</div>

<h3 style="
    color:#f87171;
    margin:0 0 9px 0;
    font-size:17px;
">
📊 Fold-Level Performance
</h3>

<p style="font-size:14px; margin:7px 0;">
For every fold, the function reports the number of validation
samples and the corresponding <b>R² score</b>. It also calculates
the <b>Mean Absolute Error</b> and the <b>Maximum Error</b> to
provide a more detailed view of prediction performance.
</p>

<p style="font-size:14px; margin:7px 0;">
The <b>R² score</b> shows how well the model explains the variation
in <b><code>Selling_Price</code></b>, while the error values show
how far the individual predictions are from their actual values.
</p>

<h3 style="
    color:#f87171;
    margin:19px 0 9px 0;
    font-size:17px;
">
🔍 Worst Prediction Analysis
</h3>

<p style="font-size:14px; margin:7px 0;">
For each fold, the absolute prediction error is calculated for
every validation sample. The observations are then sorted by
their error, and the <b>10 samples with the largest errors</b>
are selected.
</p>

<p style="font-size:14px; margin:7px 0;">
These observations are displayed together with their original
dataset information, <b>Actual</b> selling price,
<b>Predicted</b> selling price, and <b>Error</b>. This makes it
possible to investigate which vehicles are the most difficult
for the model to predict.
</p>

">
📈 Cross-Validation Summary
</h3>

<p style="font-size:14px; margin:7px 0;">
After all folds are completed, the individual R² scores are
converted into an array and summarized using the
<b>mean</b>, <b>standard deviation</b>, <b>maximum</b>, and
<b>minimum</b> values.

```python
def cross_validation_analysis(model, X, y, df, n_splits=5, shuffle=False, random_state=None, top_n=10):

    kf = KFold(
        n_splits=n_splits,
        shuffle=shuffle,
        random_state=random_state if shuffle else None
    )

    cv_scores = []

    for fold, (train_idx, val_idx) in enumerate(
        kf.split(X),
        start=1
    ):

        X_train_fold = X.iloc[train_idx]
        X_val_fold = X.iloc[val_idx]

        y_train_fold = y[train_idx]
        y_val_fold = y[val_idx]

        scaler = StandardScaler()

        X_train_fold_scaled = scaler.fit_transform(
            X_train_fold
        )

        X_val_fold_scaled = scaler.transform(
            X_val_fold
        )

        model.fit(
            X_train_fold_scaled,
            y_train_fold
        )

        y_pred_fold = model.predict(
            X_val_fold_scaled
        )

        score = r2_score(
            y_val_fold,
            y_pred_fold
        )

        cv_scores.append(score)

        original_indices = X.index[val_idx]

        fold_data = df.loc[
            original_indices
        ].copy()

        fold_data["Actual"] = (
            y_val_fold.flatten()
        )

        fold_data["Predicted"] = (
            y_pred_fold.flatten()
        )

        fold_data["Error"] = np.abs(
            y_val_fold.flatten()
            -
            y_pred_fold.flatten()
        )

        worst_samples = (
            fold_data
            .sort_values(
                by="Error",
                ascending=False
            )
            .head(top_n)
        )

        worst_indices = (
            worst_samples.index.tolist()
        )

        print("\n" + "=" * 75)

        print(
            f"Fold {fold} | "
            f"Validation Samples: {len(val_idx)} | "
            f"R² Score: {score:.4f}"
        )

        print(
            f"Mean Absolute Error: "
            f"{fold_data['Error'].mean():.4f}"
        )

        print(
            f"Maximum Error: "
            f"{fold_data['Error'].max():.4f}"
        )

        print(
            "\n10 Worst Sample Indexes:"
        )

        print(worst_indices)

        print("=" * 75)

        display(
            style_table(
                worst_samples
            )
        )

    cv_scores = np.array(cv_scores)

    print("\n" + "=" * 75)

    print("CROSS-VALIDATION SUMMARY")

    print("=" * 75)

    print(
        "CV Scores:",
        np.round(cv_scores, 4)
    )

    print(
        f"Mean CV R²: "
        f"{cv_scores.mean():.4f}"
    )

    print(
        f"Standard Deviation: "
        f"{cv_scores.std():.4f}"
    )

    print(
        f"Best Fold R²: "
        f"{cv_scores.max():.4f}"
    )

    print(
        f"Worst Fold R²: "
        f"{cv_scores.min():.4f}"
    )

    print("=" * 75)

    return cv_scores
```

```python
def find_suspicious_samples(df):

    suspicious_samples = []

    for idx, row in df.iterrows():

        selling_price = row["Selling_Price"]
        present_price = row["Present_Price"]
        kms = row["Kms_Driven"]
        age = row["Vehicle_Age"]

        reasons = []

        price_ratio = (
            selling_price / present_price
            if present_price > 0
            else np.nan
        )

        if price_ratio < 0.15:
            reasons.append("Very low price ratio")

        elif price_ratio > 1.05:
            reasons.append(
                "Selling price higher than present price"
            )

        if age > 0:

            km_per_year = kms / age

            if km_per_year > 40000:
                reasons.append("High km/year")

        else:
            km_per_year = np.nan

        if age <= 3 and kms > 100000:
            reasons.append(
                "High mileage for young vehicle"
            )

        if reasons:

            suspicious_samples.append({
                "Index": idx,
                "Selling_Price": selling_price,
                "Present_Price": present_price,
                "Kms_Driven": kms,
                "Vehicle_Age": age,
                "Price_Ratio": round(price_ratio, 3),
                "KM_per_Year": (
                    round(km_per_year, 0)
                    if not np.isnan(km_per_year)
                    else np.nan
                ),
                "Reasons": " | ".join(reasons)
            })

    suspicious_df = pd.DataFrame(suspicious_samples)

    print(
        f"Number of suspicious samples: "
        f"{len(suspicious_df)}"
    )

    display(style_table(suspicious_df))

    return suspicious_df
```

The function then creates the same engineered features used by

">
📏 Feature Scaling & Prediction
</h3>

<p style="font-size:14px; margin:7px 0;">
After creating all required features, the input columns are
reordered according to the feature order stored in
<b><code>scaler3.feature_names_in_</code></b>.
This ensures that the prediction input matches the exact structure
expected by the trained model.
</p>

<p style="font-size:14px; margin:7px 0;">
The data is then standardized using the previously fitted
<b><code>scaler3</code></b>. Finally, the scaled input is passed to
the trained <b><code>regressor3</code></b> model to generate the
predicted selling price.
</p>

">

<p style="font-size:13px; margin:0;">
🧠 <b>Important:</b><br>
The scaler is reused from the model-training stage rather than
being fitted again on the new input. This keeps the prediction
consistent with the preprocessing applied during model training.
</p>

</div>

    color:#f87171;
    margin:20px 0 9px 0;

">
🎯 Prediction Output
</h3>

<p style="font-size:14px; margin:7px 0;">
The trained model generates the predicted
<b><code>Selling_Price</code></b>, which is formatted to two decimal
places and displayed together with a clear prediction header.
The function also returns the numerical prediction so it can be
used in further calculations or analysis.
</p>

<div style="
    background:linear-gradient(135deg, #211919, #181818);
    border:1px solid #ef4444;

    margin:0;
">
🚀 <b>From Vehicle Specifications → Feature Engineering → Scaling → Final Prediction</b>
</p>

</div>

```python
def predict_car_price(present_price, kms_driven, vehicle_age, owner, fuel_type, seller_type, transmission):

    fuel_mapping = {
        "Petrol": 0,
        "Diesel": 1,
        "CNG": 2
    }

    seller_mapping = {
        "Dealer": 0,
        "Individual": 1
    }

    transmission_mapping = {
        "Manual": 0,
        "Automatic": 1
    }

    fuel_encoded = fuel_mapping[fuel_type]
    seller_encoded = seller_mapping[seller_type]
    transmission_encoded = transmission_mapping[transmission]

    input_data = pd.DataFrame({
        "Present_Price": [present_price],
        "Present_Price^2": [present_price ** 2],
        "Vehicle_Age": [vehicle_age],
        "Vehicle_Age^2": [vehicle_age ** 2],
        "Present_Price * age": [present_price * vehicle_age],
        "Present_Price * Kms_Driven": [present_price * kms_driven],
        "Present_Price * Fuel_Type" : [present_price * fuel_encoded],
        "Kms_Driven": [kms_driven],
        "Kms_Driven^2": [kms_driven ** 2],
        "Owner": [owner],
        "Fuel_Type_encoded": [fuel_encoded],
        "Seller_Type_encoded": [seller_encoded],
        "Transmission_encoded": [transmission_encoded]
    })
    input_data = input_data[scaler3.feature_names_in_]
    input_scaled = scaler3.transform(input_data)
    prediction = regressor3.predict(input_scaled)

    print("=" * 50)
    print("Vehicle Price Prediction")
    print("=" * 50)
    print(f"Predicted Selling Price: {prediction[0][0]:.2f}")
    print("=" * 50)

    return prediction[0][0]
```
